## Distribution of RA across the catalog

`map_partitions` runs the same binning function on every partition, so each worker
only ever holds one partition's worth of RA values and returns 72 counts. Summing
those per-partition counts gives the histogram for the whole catalog.

```python
import numpy as np
import pandas as pd

# Fixed 5-degree bins spanning the full RA range, so every partition returns
# counts on the same bins and they can simply be added together.
BIN_WIDTH_DEG = 5.0
RA_BIN_EDGES = np.arange(0.0, 360.0 + BIN_WIDTH_DEG, BIN_WIDTH_DEG)


def ra_histogram(partition, bin_edges):
    """Bin one partition's `ra` column, returning counts indexed by bin left edge."""
    ra = partition["ra"].to_numpy(dtype="float64", na_value=np.nan)
    counts, _ = np.histogram(ra, bins=bin_edges)
    return pd.Series(counts, index=bin_edges[:-1], name="count")


# meta for the map_partitions calls below: one integer count per bin.
HIST_META = pd.Series(dtype="int64", name="count")

# The column selection has to happen at open_catalog time, not as `gaia[["ra"]]`:
# a later selection becomes a separate `getitem` task that only runs *after*
# `read_pixel` has pulled all 152 columns into memory, which kills the workers.
# Passing `columns` here keeps the parquet reader from ever touching the rest.
gaia_ra = lsdb.open_catalog("s3://stpubdata/gaia/gaia_dr3/public/hats", columns=["ra"])
gaia_ra
```

```python
# Smoke test on 8 partitions first (the full catalog is a long read from S3)
sample_counts = (
    gaia_ra.partitions[:8]
    .map_partitions(ra_histogram, RA_BIN_EDGES, meta=HIST_META)
    .compute()
    .groupby(level=0)
    .sum()
    .sort_index()
)
print(f"{sample_counts.sum():,} sources in 8 partitions")
sample_counts.head()
```

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import EngFormatter


def plot_histogram(counts):
    fig, ax = plt.subplots(figsize=(10, 4))
    ax.stairs(counts.to_numpy(), RA_BIN_EDGES, fill=True, color="#4c78a8", edgecolor="white", linewidth=0.5)
    ax.set_xlim(0, 360)
    ax.set_xticks(np.arange(0, 361, 60))
    ax.set_xlabel("RA (degrees)")
    ax.set_ylabel(f"Sources per {BIN_WIDTH_DEG:g}° bin")
    ax.set_title("Gaia DR3 source counts by right ascension")
    ax.yaxis.set_major_formatter(EngFormatter(places=0))
    ax.grid(axis="y", color="0.9", linewidth=0.8)
    ax.set_axisbelow(True)
    for side in ("top", "right"):
        ax.spines[side].set_visible(False)
    fig.tight_layout()

plot_histogram(sample_counts)
```

```python
%%time
# Full catalog: one pass over every partition. Each partition contributes 72
# counts, so only npartitions x 72 rows come back to the client to be summed.
ra_counts = (
    gaia_ra.map_partitions(ra_histogram, RA_BIN_EDGES, meta=HIST_META)
    .compute()
    .groupby(level=0)
    .sum()
    .sort_index()
)
print(f"{ra_counts.sum():,} sources total")
ra_counts
```

```python
plot_histogram(ra_counts)
```
